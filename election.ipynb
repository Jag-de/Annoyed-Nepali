```python
df["Election"] = [
    "1981 Nepalese general election, Rastriya Panchayat",
    "1986 Nepalese general election, Rastriya Panchayat",
    "1991 Nepalese general election, House of Representatives",
    "1994 Nepalese general election, House of Representatives",
    "1999 Nepalese general election, House of Representatives",
    "2008 Nepalese Constituent Assembly election -I",
    "2013 Nepalese Constituent Assembly election-II",
    "2017 Nepalese general election, HoR",
    "2022 Nepalese general election, HoR",
    "2026 Nepalese general election, HoR"
]
```

```python
df = df.set_index("Election")
df
```

```python
#For the 1981 Nepalese general election, Rastriya Panchayat Invalid votes were 223118.(source- https://en.wikipedia.org/wiki/1981_Nepalese_general_election)
#For the 2008 Nepalese Constituent Assembly election-I Registration were  17611832.(source-https://en.wikipedia.org/wiki/2008_Nepalese_Constituent_Assembly_election
#)
```

```python
df.loc["1981 Nepalese general election, Rastriya Panchayat", "Invalid votes"] = "5.47%"
df.loc["2008 Nepalese Constituent Assembly election -I", "Registration"] = "17611832"
df
```

```python
#converting str to numericals format so calcuations can be operated.
```

```python
per_cols = ['Voter Turnout', 'VAP Turnout', 'Invalid votes']
num_cols = ['Total vote', 'Voting age population','Registration', 'Population']

df[per_cols] = df[per_cols].replace('%', '', regex=True).astype(float)
df[num_cols] = df[num_cols].replace(',', '', regex=True).astype('Int64')
df
```

```python
df.dtypes
```

```python
plt.figure(figsize=(12, 6))

plt.bar(df["Year"], df["VAP Turnout"])

plt.xlabel("Year")
plt.ylabel("VAP Turnout (%)")
plt.title("VAP Turnout Over the Years")

plt.ylim(0, 100)
plt.xticks(rotation=45)

plt.tight_layout()
plt.show()
```

```python
df_plot = df[["Year", "Voter Turnout", "VAP Turnout"]]
df_plot = df_plot.melt(id_vars="Year",var_name="Turnout Type",value_name="Turnout")

plt.figure(figsize=(12, 6))
sns.barplot(data=df_plot,x="Year",y="Turnout",hue="Turnout Type")
plt.title("VAP Turnout vs Voter Turnout Over the Elections")
plt.xlabel("Year")
plt.ylabel("Turnout (%)")
plt.xticks(rotation=45)
plt.legend(title="Turnout Type")
plt.tight_layout()
plt.show()
```
